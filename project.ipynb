```python
import pandas as pd

# Load dataset to keep df available
df = pd.read_excel("videogamesSales.xlsx")

# Define variables
game_name = df["Name"].iloc[0]
game_year = df["Year"].iloc[0]
global_sales = df["Global_Sales"].iloc[0]

# Checking the data types of our variables
print("Data types:")
print("Game name:", type(game_name))
print("Release year:", type(game_year))
print("Global sales:", type(global_sales))
```

```python
# Different types of variables

total_games = len(df)                        
average_sales = df["Global_Sales"].mean()    
dataset_name = "Video Game Sales"            
has_missing = df.isnull().sum().sum() > 0    

print("Total games:", total_games)
print("Average sales:", average_sales)
print("Dataset name:", dataset_name)
print("Has missing values:", has_missing)
```

```python
# Lists and dictionaries

regions = ["NA_Sales", "EU_Sales", "JP_Sales", "Other_Sales"]

top_game = {
    "Name": "Wii Sports",
    "Sales": 82.74
}

print("Sales regions:", regions)
print("Top game:", top_game["Name"])
print("Top game sales:", top_game["Sales"])
```

```python
# Function using conditionals

def check_sales(sales):
    if sales >= 20:
        return "Blockbuster"
    elif sales >= 5:
        return "Hit"
    elif sales >= 1:
        return "Average"
    else:
        return "Low"

print(check_sales(25))
print(check_sales(8))
print(check_sales(2))
print(check_sales(0.5))
```

```python
# Classify all games by global sales

df["Sales_Category"] = df["Global_Sales"].apply(check_sales)

print(df[["Name", "Global_Sales", "Sales_Category"]].head(10))
```

```python
# Count games in each sales category

category_counts = df["Sales_Category"].value_counts()

print(category_counts)
```

```python
# For loop to display sales by region

regions = ["NA_Sales", "EU_Sales", "JP_Sales", "Other_Sales"]

for region in regions:
    total_sales = df[region].sum()
    print(region, ":", total_sales, "million")
```

```python
# For loop with condition and break

for i in range(len(df)):
    if df.loc[i, "Global_Sales"] >= 20:
        print("First blockbuster game:", df.loc[i, "Name"])
        print("Global sales:", df.loc[i, "Global_Sales"], "million")
        break
```

```python
# While loop to display first 3 games

i = 0

while i < 3:
    print(df.loc[i, "Name"])
    i = i + 1
```

```python
# Function to find top games by genre

def top_n_by_genre(data, genre, n=5):
    result = data[data["Genre"] == genre]
    result = result.sort_values("Global_Sales", ascending=False)
    return result[["Name", "Global_Sales"]].head(n)

print(top_n_by_genre(df, "Action"))
```

```python
# Function to calculate regional sales share

def region_share(data):
    regions = ["NA_Sales", "EU_Sales", "JP_Sales", "Other_Sales"]
    total_sales = data[regions].sum().sum()

    for region in regions:
        sales = data[region].sum()
        share = (sales / total_sales) * 100
        print(region, ":", round(share, 2), "%")
        
region_share(df)
```

```python
# Information about the dataset

df.info()
```

```python
# Statistical summary of the dataset

df.describe()
```

```python
# Check missing values

missing_values = df.isnull().sum()

print(missing_values)
```

```python
# Clean the dataset

df_clean = df.dropna(subset=["Year"]).copy()

df_clean["Publisher"] = df_clean["Publisher"].fillna("Unknown")

df_clean["Year"] = df_clean["Year"].astype(int)

print("Original rows:", len(df))
print("Cleaned rows:", len(df_clean))
print("Missing Year:", df_clean["Year"].isnull().sum())
print("Missing Publisher:", df_clean["Publisher"].isnull().sum())
```

```python
# Filter Action genre games

action_games = df_clean[df_clean["Genre"] == "Action"]

print("Number of Action games:", len(action_games))
print(action_games[["Name", "Platform", "Global_Sales"]].head(10))
```

```python
# Total global sales by genre

genre_sales = df_clean.groupby("Genre")["Global_Sales"].sum()

print(genre_sales.sort_values(ascending=False))
```

```python
# NumPy statistical analysis

import numpy as np

sales = df_clean["Global_Sales"]

print("Mean:", np.mean(sales))
print("Median:", np.median(sales))
print("Standard deviation:", np.std(sales))
print("Maximum sales:", np.max(sales))
print("90th percentile:", np.percentile(sales, 90))
```

```python
# Bar chart of global sales by genre

import matplotlib.pyplot as plt

genre_sales = df_clean.groupby("Genre")["Global_Sales"].sum()
genre_sales = genre_sales.sort_values(ascending=False)

plt.figure(figsize=(10, 5))
plt.bar(genre_sales.index, genre_sales.values)

plt.title("Global Video Game Sales by Genre")
plt.xlabel("Genre")
plt.ylabel("Global Sales (millions)")
plt.xticks(rotation=45)

plt.show()
```

```python
# Line chart of global sales by year

yearly_sales = df_clean.groupby("Year")["Global_Sales"].sum()

plt.figure(figsize=(10, 5))
plt.plot(yearly_sales.index, yearly_sales.values)

plt.title("Global Video Game Sales by Year")
plt.xlabel("Year")
plt.ylabel("Global Sales (millions)")

plt.show()
```

```python
# Top 10 platforms by global sales

import seaborn as sns

platform_sales = df_clean.groupby("Platform")["Global_Sales"].sum()
platform_sales = platform_sales.sort_values(ascending=False).head(10)

plt.figure(figsize=(10, 5))
sns.barplot(x=platform_sales.values, y=platform_sales.index)

plt.title("Top 10 Platforms by Global Sales")
plt.xlabel("Global Sales (millions)")
plt.ylabel("Platform")

plt.show()
```

```python
# Correlation heatmap of sales

sales_columns = [
    "NA_Sales",
    "EU_Sales",
    "JP_Sales",
    "Other_Sales",
    "Global_Sales"
]

correlation = df_clean[sales_columns].corr()

plt.figure(figsize=(8, 6))
sns.heatmap(correlation, annot=True, cmap="coolwarm")

plt.title("Correlation Between Regional and Global Sales")

plt.show()
```

```python
# Regional sales comparison

regions = ["NA_Sales", "EU_Sales", "JP_Sales", "Other_Sales"]

regional_sales = df_clean[regions].sum()

plt.figure(figsize=(8, 5))
sns.barplot(x=regional_sales.index, y=regional_sales.values)

plt.title("Video Game Sales by Region")
plt.xlabel("Region")
plt.ylabel("Sales (millions)")

plt.show()
```

```python
# Top 10 best-selling games

top_10_games = df_clean.sort_values(
    "Global_Sales", ascending=False
).head(10)

print(top_10_games[["Name", "Platform", "Genre", "Global_Sales"]])
```

```python
# Top 10 platforms by number of games

top_platforms = df_clean["Platform"].value_counts().head(10)

print(top_platforms)
```

```python
# Top 10 publishers by number of games

top_publishers = df_clean["Publisher"].value_counts().head(10)

print(top_publishers)
```

```python
# Top 10 publishers by global sales

publisher_sales = df_clean.groupby("Publisher")["Global_Sales"].sum()

publisher_sales = publisher_sales.sort_values(
    ascending=False
).head(10)

print(publisher_sales)
```

```python
# Find the year with the highest global sales

yearly_sales = df_clean.groupby("Year")["Global_Sales"].sum()

peak_year = yearly_sales.idxmax()
peak_sales = yearly_sales.max()

print("Peak sales year:", peak_year)
print("Global sales in that year:", peak_sales, "million")
```

```python
# Highest-selling game in each genre

top_games_by_genre = df_clean.loc[
    df_clean.groupby("Genre")["Global_Sales"].idxmax()
]

print(top_games_by_genre[["Genre", "Name", "Global_Sales"]].sort_values("Genre"))
```

```python
# Total sales by region

regional_sales = df_clean[
    ["NA_Sales", "EU_Sales", "JP_Sales", "Other_Sales"]
].sum()

print(regional_sales)
```

```python
# Check for duplicate rows

duplicate_rows = df_clean.duplicated().sum()

print("Number of duplicate rows:", duplicate_rows)
```

```python
# Final dataset overview

print("Rows:", len(df_clean))
print("Columns:", len(df_clean.columns))
print("Duplicate rows:", df_clean.duplicated().sum())
print("Missing values:", df_clean.isnull().sum().sum())
```

```python
# Main analysis results

top_genre = df_clean.groupby("Genre")["Global_Sales"].sum().idxmax()
top_platform = df_clean.groupby("Platform")["Global_Sales"].sum().idxmax()
top_publisher = df_clean.groupby("Publisher")["Global_Sales"].sum().idxmax()

top_game = df_clean.loc[
    df_clean["Global_Sales"].idxmax(), "Name"
]

top_game_sales = df_clean["Global_Sales"].max()

top_region = regional_sales.idxmax()

print("Top genre:", top_genre)
print("Top platform:", top_platform)
print("Top publisher:", top_publisher)
print("Top-selling game:", top_game)
print("Top game sales:", top_game_sales, "million")
print("Top region:", top_region)
print("Peak sales year:", peak_year)
```

```python
# Correlation table for sales

sales_correlation = df_clean[
    ["NA_Sales", "EU_Sales", "JP_Sales", "Other_Sales", "Global_Sales"]
].corr()

print(sales_correlation)
```

```python

# Save the cleaned dataset

df_clean.to_excel("videogamesSales_cleaned.xlsx", index=False)

print("Cleaned dataset saved successfully.")
```
